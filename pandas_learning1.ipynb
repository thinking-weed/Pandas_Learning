```python
import pandas as pd
data = [
        ["ノートPC_A","PC","120000",2],
        ["マウス","周辺機器","3000",5],
        ["キーボード","周辺機器","8000",3],
        ["モニター","周辺機器","35000",2],
        ["ノートPC_B","PC","130000",1]
        ]
col = ["商品名","カテゴリ","単価","数量"]

#indexが１始まりになるように再設定
idx = range(1,len(data)+1) 
df = pd.DataFrame(data=data, columns=col, index=idx)
df
```

```python
df = df.drop(columns=["カテゴリ"])
df
```
